```python
import pandas as pd
import plotly.express as px
from IPython.display import display

# Sample data
df = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    "Sales": [120, 150, 180, 140, 220, 260],
    "Profit": [20, 30, 35, 25, 50, 65]
})

# Create sales chart
fig = px.bar(
    df,
    x="Month",
    y="Sales",
    title="Monthly Sales Dashboard",
    labels={"Sales": "Sales ($)", "Month": "Month"}
)

fig.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.plot(
    ["Jan", "Feb", "Mar", "Apr", "May", "Jun"],
    [120, 150, 180, 140, 220, 260],
    marker="o"
)
plt.title("Recession Report Graphs")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid(True)

plt.savefig("recessionreportgraphs.png", dpi=300, bbox_inches="tight")
plt.show()
```
