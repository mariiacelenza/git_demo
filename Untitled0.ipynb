```python
import yfinance as yf
import pandas as pd
import requests
import matplotlib.pyplot as plt
from bs4 import BeautifulSoup
# Question 1: Tesla stock data
tesla = yf.Ticker("TSLA")
tesla_stock_data = tesla.history(period="max")
print("Tesla Stock Data")
display(tesla_stock_data.head())
# Question 2: Tesla revenue data (web scraping)
url_tesla = "https://www.macrotrends.net/stocks/charts/TSLA/tesla/revenue"
headers = {"User-Agent": "Mozilla/5.0"}

response = requests.get(url_tesla, headers=headers)
tesla_soup = BeautifulSoup(response.text, "html.parser")

tesla_tables = pd.read_html(response.text)
tesla_revenue = next(
    (table for table in tesla_tables
     if any("revenue" in str(col).lower() for col in table.columns)),
    pd.DataFrame()
)
print("Tesla Revenue Data")
display(tesla_revenue.tail())

# Question 3: GameStop stock data
gme = yf.Ticker("GME")
gme_stock_data = gme.history(period="max")
print("GameStop Stock Data")
display(gme_stock_data.head())

# Question 4: GameStop revenue data (web scraping)
url_gme = "https://www.macrotrends.net/stocks/charts/GME/gamestop/revenue"

response = requests.get(url_gme, headers=headers)
gme_soup = BeautifulSoup(response.text, "html.parser")

gme_tables = pd.read_html(response.text)
gme_revenue = next(
    (table for table in gme_tables
     if any("revenue" in str(col).lower() for col in table.columns)),
    pd.DataFrame()
)
print("GameStop Revenue Data")
display(gme_revenue.tail())

# Define a function that makes a graph
def make_graph(stock_data, title):
    plt.figure(figsize=(12, 5))
    plt.plot(stock_data.index, stock_data["Close"])
    plt.title(title)
    plt.xlabel("Date")
    plt.ylabel("Share Price (USD)")
    plt.grid(True)
    plt.show()

# Question 5: Plot Tesla stock graph
make_graph(tesla_stock_data, "Tesla Stock Price Over Time")

# Question 6: Plot GameStop stock graph
make_graph(gme_stock_data, "GameStop Stock Price Over Time")

```
